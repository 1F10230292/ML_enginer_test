課題説明資料および[ETDataset](https://github.com/zhouhaoyi/ETDataset/tree/main)の背景を踏まえ、PoC（技術検証）として「オイル温度の将来予測が運用・保全にどう貢献できるか」を示すために置くべき主要な仮説は以下の通りです。

**1. 予測精度と異常検知（予防保全）に関する仮説**

* **仮説**: 「電力負荷（HUFL〜LULL）や過去の温度推移の時系列パターンを学習することで、閾値超過（危険温度）が発生する数時間〜数日前に高精度な予兆予測が可能である。」
* **背景**: 従来の単なる「今の温度に対する閾値監視（事後対応）」から「将来の温度予測（事前検知）」へ移行することで、変圧器の過熱や事故を未然に防ぐメンテナンス計画が立てられるようになります。

**2. ドメイン知識・データ構造に関する仮説**

* **仮説**: 「オイル温度（OT）の変化には物理的な熱慣性（遅延）が存在するため、現在の電力負荷の上昇は即座ではなく一定の時間差（ラグ）をもってOTに影響する。」
* **背景**: 外生変数（電力負荷）のラグ特徴量や、日周期・週周期の季節性をモデルに含めることで、単なる自己回帰モデル（過去のOTのみを使うモデル）よりも予測精度が向上します。

**3. ビジネス価値・運用インパクトに関する仮説**

* **仮説**: 「事前予測に基づいて変圧器の負荷調整や冷却運用を行うことで、設備寿命の延伸および不要な安全マージン（超過抑止のための過剰な出力制限）の削減が可能になる。」
* **背景**: 予兆検知により突然の停電リスクを低減すると同時に、過度なマージン確保による電力制御の無駄を排除できるため、運用効率化の定量的なインパクトが示せます。

**4. データ解像度（15分刻み vs 1時間刻み）に関する仮説**

* **仮説**: 「1時間刻み（h1/h2）データでも長期的・巨視的な温度トレンドは捕捉できるが、急激なスパイク（瞬間的なスパイク過熱）を捉えるには15分刻み（m1/m2）データによる高解像度予測が有効である。」
* **背景**: 計算コストと予測精度のトレードオフを検証・検討する際の軸となります。

変圧器の保全・設備劣化において、オイル温度（OT）が「壊れるリスク（故障・劣化リスク）」に直結する理由と、具体的な閾値（しきい値）の考え方は以下の通りです。

---

### **1. 壊れるリスクを示すための指標（閾値の考え方）**

変圧器の故障リスクは、主に「絶縁油・絶縁紙（巻線を覆う紙）の熱劣化」によって発生します。業界標準（IEC 60076-7 や IEEE C57.91 などの変圧器負荷ガイド）に基づくと、温度の基準値（閾値）は以下のように定義されています。

* **正常動作領域（〜65℃〜80℃）**:
* 通常運用時の上限値。変圧器の設計寿命（約20〜30年）を維持できる標準的な温度帯です。


* **警戒領域（85℃〜95℃）**:
* 絶縁材料（特に紙）の**経年劣化が加速**し始める閾値です。一般に「油温が6℃〜8℃上昇するごとに、絶縁材料の寿命（熱劣化速度）は2倍になる（アレニウスの法則）」と言われています。


* **危険・異常領域（105℃〜115℃以上）**:
* **緊急対応が必要な臨界閾値**です。105℃を超えると絶縁油からガスが発生（分解ガス）しやすくなり、115℃を超えると急速な絶縁破壊・引火・内部短縮事故（ショート）を引き起こすリスクが跳ね上がります。



※データセット（ETDataset）の数値を見る際、温度軸の単位（℃など）や標準化の有無を確認しつつ、上記のような「物理的な危険閾値」をマッピングしてリスクを可視化します。

---

### **2. 機械学習モデルで「壊れるリスク」を可視化・定量化する3つの方法**

課題で「事後対応から予測による予防保全へ移行したい」ことを示すには、単に温度の数字を当てるだけでなく、**リスクの可視化**まで踏み込んだアプローチが有効です。

* **① 「危険閾値への到達予測時間（Time-To-Threshold）」の提示**
* **アプローチ**: 今後の負荷推移から「あと何時間後に85℃（警戒）または105℃（危険）に到達するか」を予測します。
* **価値**: 「12時間後に90℃に達する見込みです」と提示できれば、運用者は「冷却ファンの追加作動」や「他系統への負荷切替」を事前に行うことができます。


* **② 「累積熱ストレス（寿命消費率）」の可視化**
* **アプローチ**: アレニウスの法則に基づき、予測されたオイル温度推移から「このままだと本日の運用で何日分（何時間分）の設備寿命が縮まるか」を計算します。
* **価値**: 一時的な超過だけでなく、長時間の高負荷運用による「目に見えない設備の消耗（累積リスク）」を定量化できます。


* **③ 「異常乖離（モデル予測値 vs 実際の温度）」による初期異常の検知**
* **アプローチ**: 負荷（HUFL等）から予測される「本来あるべき正常温度」に対し、実際の温度が明らかに高く推移し始めた場合、それを「冷却系の目詰まりや油劣化のサイン」と判定します。
* **価値**: 突然の破断事故につながる前に、設備の潜在的な異常を事前アラートできます。

### **PoC報告スライドへの盛り込み方の提案**

スライド内で「壊れるリスクと予測の価値」を主張する際は、以下のような文脈でまとめるのが効果的です。

> **【メッセージ】**
> 油温の上昇は絶縁寿命を指数関数的に縮めるため、事後アラート（閾値超過）ではなく「数時間後の閾値到達予測」によって事前に対策を打つ予防保全が不可欠である。
> **【ボディ（裏付け）】**
> * 85℃を超えるとアレニウス則により劣化速度が急速に加速する。
> * モデルにより未来のOTを予測することで、「危険閾値（例: 85℃/105℃）に達する前の事前回避運用（負荷制限・冷却強化）」が可能となり、突発停止リスクの解消と設備寿命の延伸に寄与する。
